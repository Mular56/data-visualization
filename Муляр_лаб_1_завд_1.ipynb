```python
'''1. Побудувати рекурентну нейронну мережу для прогнозування слів.
 Підібрати архітектуру моделі, кількість  і тип рекурентних шарів, гіперпараметри моделі.
 Побудувати графіки функції втрат і точності.
 Також доцільно викоритстати Керас тюнер для підбору параметрів. '''

import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.layers import Dense, GRU, LSTM, Embedding, Dropout, Bidirectional
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical

# Текст для навчання
texts = """
На панщині пшеницю жала,
Втомилася; не спочивать
Пішла в снопи, пошкандибала
Івана сина годувать.

Воно сповитеє кричало
У холодочку за снопом.
Розповила, нагодувала,
Попестила; і ніби сном,

Над сином сидя, задрімала.
І сниться ій той син Іван
І уродливий, і багатий,
Не одинокий, а жонатий
"""

# Попередня обробка тексту
texts = texts.replace('\ufeff', '')
maxWordsCount = 1000
tokenizer = Tokenizer(num_words=maxWordsCount, filters='1-"-#$%&; ()*+,-./:; <=>?@[\\]^_{}}~\t\n\r',
                      lower=True, split=' ', char_level=False)
tokenizer. fit_on_texts([texts])

#Перетворюємо текст у числову послідовність
data = tokenizer.texts_to_sequences ([texts])
res = np.array(data[0])

inp_words = 3 # Кількість слів у вхідній послідовності
n = res.shape[0] - inp_words

# Формуємо навчальні вибірки
X = np.array([res[i:i + inp_words] for i in range(n)])
Y = to_categorical (res [inp_words: ], num_classes=maxWordsCount)

# Побудова покращеної моделі з Bidirectional LSTM та GRU
model = Sequential()
model.add(Embedding(maxWordsCount, 256, input_length=inp_words)) # Beкторизачiя слiв
model.add(Bidirectional(LSTM(128, activation='tanh', return_sequences=True))) # Bidirectional LSTM
model.add(GRU(128, activation='tanh')) # Додаємо GRU-шар
model.add(Dropout(0.3)) # Dropout для регуляризації
model.add(Dense(256, activation='relu')) # Повнозв'язний шар
model.add(Dense(maxWordsCount, activation='softmax')) # Вихідний шар

# Компільовуємо модель
model.compile(loss='categorical_crossentropy', metrics=['accuracy'], optimizer='adam')


# Навчання моделі
history = model.fit(X, Y, batch_size=32, epochs=50)

# Графіки функції втрат та точності
plt.figure(figsize=(12, 4))

# Графік втрат
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Графік функції втрат')

# Графік точності
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Accuracy', color='green')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Графік точності')

plt.show()

# Вивід архітектури моделі
model.summary()

# Функція для генерації тексту
def buildPhrase (texts, str_len=20):
    res = texts
    data = tokenizer.texts_to_sequences ([texts])[0]
    for i in range(str_len):
        x = data[i: i + inp_words]
        inp = np.expand_dims(x, axis=0)

        pred = model.predict(inp)
        indx = pred.argmax(axis=1)[0]
        data.append(indx)

        res += " " + tokenizer.index_word.get(indx, '') #Додамо нове слово

    return res

#Генерація нового тексту
res = buildPhrase("На панщині пшеницю")
print(res)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import keras_tuner as kt
from tensorflow.keras.layers import Dense, GRU, LSTM, Embedding, Dropout, Bidirectional, SimpleRNN
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical

# Текст для навчання
texts = """
На панщині пшеницю жала,
Втомилася; не спочивать
Пішла в снопи, пошкандибала
Івана сина годувать.

Воно сповитеє кричало
У холодочку за снопом.
Розповила, нагодувала,
Попестила; і ніби сном,

Над сином сидя, задрімала.
І сниться ій той син Іван
І уродливий, і багатий,
Не одинокий, а жонатий
"""

# Попередня обробка тексту
texts = texts.replace('\ufeff', '')
maxWordsCount = 1000
tokenizer = Tokenizer(num_words=maxWordsCount, filters='1-"-#$%&; ()*+,-./:; <=>?@[\\]^_{}}~\t\n\r',
                      lower=True, split=' ', char_level=False)
tokenizer. fit_on_texts([texts])

#Перетворюємо текст у числову послідовність
data = tokenizer.texts_to_sequences ([texts])
res = np.array(data[0])

inp_words = 3 # Кількість слів у вхідній послідовності
n = res.shape[0] - inp_words

# Формуємо навчальні вибірки
X = np.array([res[i:i + inp_words] for i in range(n)])
Y = to_categorical (res [inp_words: ], num_classes=maxWordsCount)


# Функція для створення моделі (HyperModel)
def build_model(hp):
    model = Sequential()
    model.add(Embedding (maxWordsCount, hp. Int('embedding_dim', 128, 512, step=64), input_length=inp_words))
    model.add(SimpleRNN(hp.Int('rnn_units', 64, 256, step=64), activation= 'tanh', return_sequences=True))
    model.add(LSTM(hp.Int('1stm_units', 64, 256, step=64), activation= 'tanh'))
    model.add(Dropout (hp. Float('dropout_rate', 0.1, 0.5, step=0.1)))
    model.add(Dense (hp. Int('dense_units', 64, 256, step=64), activation= 'relu'))
    model.add(Dense (maxWordsCount, activation='softmax'))

    model.compile(
    loss='categorical_crossentropy',
    metrics=['accuracy'],
    optimizer=hp.Choice('optimizer', ['adam', 'rmsprop'])
    )
    return model


# Використання Keras Tuner
tuner = kt.RandomSearch(
    build_model,
    objective='accuracy',
    max_trials=10,
    executions_per_trial=1,
    directory='my_tuning',
    project_name='text_prediction'
)
# Запуск підбору параметрів
tuner.search(X, Y, epochs=10, batch_size=32)

# Отримуємо найкращі параметри
best_hps = tuner.get_best_hyperparameters (num_trials=1)[0]
print(f"""
Найкращі параметри:
Embedding Dimension: {best_hps.get('embedding_dim')}
RNN Units: {best_hps.get('rnn_units')}
LSTM Units: {best_hps.get('1stm_units')}
Dropout Rate: {best_hps.get('dropout_rate')}
Dense Units: {best_hps.get('dense_units')}
Optimizer: {best_hps.get('optimizer')} """)

# Будуємо модель з найкращими параметрами
best_model = tuner.hypermodel.build(best_hps)

# Навчаємо модель з найкращими параметрами
history = best_model.fit(X, Y, batch_size=32, epochs=50)


# Графіки функції втрат та точності
plt.figure(figsize=(12, 4))

# Графік втрат
plt.subplot(1, 2, 1)
plt.plot (history.history['loss'], label='Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Графік функції втрат')


# Графік точності
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Accuracy', color='green')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Графік точності')

plt.show()

# Вивід архітектури моделі
best_model.summary()

# Функція для генерації тексту
def buildPhrase (texts, str_len=20):
    res = texts
    data = tokenizer.texts_to_sequences([texts])[0]
    for i in range(str_len):
        x = data[i: i + inp_words]
        inp = np.expand_dims (x, axis=0)

        pred = best_model.predict(inp)
        indx= pred. argmax(axis=1)[0]
        data.append(indx)

        res += " " + tokenizer.index_word.get(indx, '') # ДодаЄмо нове слово
    return res

#Генерація нового тексту
res = buildPhrase("На панщині пшеницю")
print(res)
```
